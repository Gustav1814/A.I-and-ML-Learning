```python
import torch
import torch.nn as nn
import torch.nn.functional as F
```

**Create a Model Class That Inherits nn.Module**

```python
class Model(nn.Module):
  # Input layer (4 Feature of the flower)-----> Hidden Layer 1 (nuerons)-------> Hidden Layer 2(Nuerons)--------> Output (3 classes of iris flower)
  def __init__(self, in_features=4, h1=8, h2=9, out_features=3):
    super().__init__()
    self.fc1 = nn.Linear(in_features, h1)
    self.fc2 = nn.Linear(h1, h2)
    self.out = nn.Linear(h2, out_features)

  def forward(self,x):
    x = F.relu(self, fc1(x))   # Rectified Linear Unit
    x = F.relu(self, fc2(x))
    x = self.out(x)
    return x

```

```python
# Pick a manual seed for randomization
torch.manual_seed(41)

# Create an Instance of Model
model = Model()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
url = 'https://gist.githubusercontent.com/curran/a08a1080b88344b0c8a7/raw/0e7a9b0a5d22642a06d3d5b9bcbad9890c8ee534/iris.csv'
my_df = pd.read_csv(url)
```

```python
my_df.tail()
```

```python
# Changed Last Column From String to Integer
my_df['species']= my_df['species'].replace('setosa', 0.0)
my_df['species']= my_df['species'].replace('versicolor', 1.0)
my_df['species']= my_df['species'].replace('virginica', 2.0)
my_df
```

**Train Test Split**

```python
X = my_df.drop('species', axis =1)
Y = my_df['species']
```

```python
X = X.values
Y = Y.values
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=41)
```

```python
x_train = torch.FloatTensor(x_train)
x_test = torch.FloatTensor(x_test)
y_train = torch.LongTensor(y_train)
y_test = torch.LongTensor(y_test)
```

**Set Criterion of Model to Measure the Error**

```python
criterion = nn.CrossEntropyLoss()
```

**Choose Optimizer**

```python
# Adam Optimizer and learning Rate (If error doesnt go down after bunch of iterations (epochs), lower our learning rate)
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)
```

**TRAIN OUR MODEL**


*   Epochs? (One Run thru all the training data in our network)




```python
epochs = 100
```

```python
losses = []
for i in range(epochs):
  # Go Forward and get a prediction
  y_pred = model.forward(x_train) # Get predicted results

  # Measure the loss/error, gonna be high at first
  loss = criterion(y_pred, y_train) # Predicted values vs the y_train

  # Keep Track of our losses
  losses.append(loss.detach().numpy())

  # Print every 10 epoch

  if i % 10 == 0:
    print(f'Epoch: {i} and loss: {loss}')


  # Do some back propagation: take the error rate of forward propagation and feed it back thru the network to fine tune the weights
  optimizer.zero_grad()
  loss.backward()
  optimizer.step()

```
