```python
import torch
```

```python
mytorch = torch.arange(20)
mytorch
```

**RESHAPE AND VIEW**

```python
mytorch = mytorch.reshape(4,5)
mytorch
```

**Reshape when we Dont know the number of items using -1**

```python
mytorch2 = torch.arange(10)
mytorch2
```

```python
mytorch2 = mytorch2.reshape (2,-1)
mytorch2
```

```python
mytorch2 = mytorch2.reshape (-1,5)
mytorch2
```

```python
mytorch3 = torch.arange(15)
mytorch3
```



> using View



```python
mytoorch4 = mytorch3.view(3,5)
mytoorch4
```
