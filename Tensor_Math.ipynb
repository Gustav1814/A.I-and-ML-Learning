## **TENSOR MATH OPERATIONS**


*   ADD, SUBTRACT, MULTIPLY, DIVIDE, REMIANDER, EXPONENTS
*   SHORTHAND AND LONGHAND

*   REASSINGMENT





```python
import torch
import numpy as np
```

```python
tensor_A = torch.tensor([1,2,3,4])
tensor_B = torch.tensor([5,6,7,8])
```

**Addition**

```python
tensor_A + tensor_B
```



> Add longhand



```python
torch.add(tensor_A, tensor_B)
```

**Subtraction**

```python
tensor_B - tensor_A
```

```python
torch.subtract(tensor_B, tensor_A)
```

**Multiplication**

```python
tensor_A * tensor_B
```

```python
torch.mul(tensor_A, tensor_B)
```

**Division**

```python
tensor_B / tensor_A
```

```python
torch.div(tensor_B, tensor_A)
```

**Remainder Modulus**

```python
tensor_B % tensor_A
```

```python
torch.remainder(tensor_B, tensor_A)
```

**Exponents**

```python
torch.pow(tensor_A, tensor_B)
```

**Reassignment**

```python
tensor_A
```

```python
tensor_A - tensor_A + tensor_B
```

```python
tensor_A.add_(tensor_B)
```
